```python
VP = 80
VN = 70
FP = 10
FN = 20

N = VP + VN + FP + FN
print(f"Matriz de confusão usada: VP={VP}, VN={VN}, FP={FP}, FN={FN}")
print(f"Total N = {N}")
```

```python
def sensibilidade(vp, fn):
  return vp / (vp + fn)

def especificidade(vn, fp):
  return vn / (fp + vn)

def acuracia(vp, vn, n):
  return (vp + vn) /n

def precisao(vp, fp):
  return vp / (vp + fp)

def f_score(p, s):
  if (p + s) == 0:
    return 0
  return 2 * (p * s) / (p + s)
```

```python
S = sensibilidade(VP, FN)
E = especificidade(VN, FP)
A = acuracia(VP, VN, N)
P = precisao(VP, FP)
F = f_score(P, S)

print(f"Sensibilidade (Recall): {S:.4f}")
print(f"Especificidade: {E:.4f}")
print(f"Acurácia: {A:.4f}")
print(f"Precisão: {P:.4f}")
print(f"F-Score: {F:.4f}")
```
